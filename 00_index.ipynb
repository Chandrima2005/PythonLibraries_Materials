## How to use these notebooks

- Read the chapters in order the first time. Each one builds on the ones before it and says which.
- Run every code cell yourself and change the numbers. Breaking things on purpose is the fastest way to learn.
- Every chapter ends with a **Common mistakes** table, a **Cheat sheet** and **Exercises** with solutions. Try each exercise before you look at the solution.
- The notebooks expect to run from the repository root, so that `data/orders.csv` can be found.

## The dataset

Most examples use one realistic dataset: a year (2024) of sales at **Brew & Byte**, a fictional chain of four coffee shops in Pune, Mumbai and Bengaluru. It is generated by `tools/make_data.py`, so it is free to use and identical on every machine.

| File | Rows | What is in it |
|---|---|---|
| `orders.csv` | ~53,000 | one row per item ordered: time, store, customer, item, quantity, price, payment |
| `customers.csv` | 600 | loyalty-card customers: age, home store, joining date, tier |
| `stores.csv` | 4 | city, opening year and seats for each store |
| `menu.csv` | 10 | items, categories and prices |
| `weather.csv` | ~1,100 | daily temperature and rainfall for each city |
| `orders_messy.csv` | ~1,500 | a deliberately dirty sample for the cleaning chapter |
